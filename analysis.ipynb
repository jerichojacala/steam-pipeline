# Azure Storage — Preview Unstructured Data

Query the first rows of a file (or folder) in Azure Blob / ADLS Gen2 storage using `read_files()`.

```python
%sql
-- Replace the placeholders below with your actual values:
--   <container>       : your blob container / filesystem name
--   <storage_account> : your Azure storage account name
--   <path/to/file>    : path to your file or folder (e.g. raw/logs/2024/)
--   format            : auto-detected, or specify 'csv', 'json', 'parquet', 'text', 'avro', etc.

SELECT *
FROM read_files(
  'abfss://steamcontainer@jjacalasteamstorage.dfs.core.windows.net/bronze/extended_steam_games.csv',
  format => 'csv',
  header => 'true'
  -- uncomment to add schema: schema => 'col1 TYPE, col2 TYPE, ...'
)
LIMIT 10
```

```python
# Read games.json from Azure Storage and write it as CSV
df = spark.read.option("multiline", "true").json(
    "abfss://steamcontainer@jjacalasteamstorage.dfs.core.windows.net/bronze/games.json"
)

# Write to CSV in the same bronze container
df.write.mode("overwrite").option("header", "true").csv(
    "abfss://steamcontainer@jjacalasteamstorage.dfs.core.windows.net/silver/games_csv"
)

print(f"Done! Rows written: {df.count()}")
print(f"Columns: {df.columns}")
```
